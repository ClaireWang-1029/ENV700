````python
# Write your solution here
import math
#A
def circle_area(r):
    area= math.pi*r**2
    return area
circle_area(5)
````

````python
#B
def rectangle_area(length,width):
    area= length * width
    return area
rectangle_area(8,6)
````

````python
def min_max_values(numbers):
    minimum=min(numbers)
    maximum=max(numbers)
    return minimum,maximum
numbers =[3,5,7,10]
min_max_values(numbers)
````
